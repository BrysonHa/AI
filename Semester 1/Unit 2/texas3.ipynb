# When drawing 2 cards at a time from a standard deck of cards, what is the probability of drawing a pair 2 times in a row? Assume no cards have been drawn initially.

This study will be a prospective, observational study. The sample space I am observing is all of the cards in a standard deck so 52. One trial of the simulation is the dealing of four pairs of cards. I am measuring instances of a pair being dealt twice in a row.

```python
import random, seaborn
```

```python
def genDeck():
    deck = [
        ("2", "Hearts"), ("3", "Hearts"), ("4", "Hearts"), ("5", "Hearts"), ("6", "Hearts"),
        ("7", "Hearts"), ("8", "Hearts"), ("9", "Hearts"), ("10", "Hearts"),
        ("Jack", "Hearts"), ("Queen", "Hearts"), ("King", "Hearts"), ("Ace", "Hearts"),
    
        ("2", "Diamonds"), ("3", "Diamonds"), ("4", "Diamonds"), ("5", "Diamonds"), ("6", "Diamonds"),
        ("7", "Diamonds"), ("8", "Diamonds"), ("9", "Diamonds"), ("10", "Diamonds"),
        ("Jack", "Diamonds"), ("Queen", "Diamonds"), ("King", "Diamonds"), ("Ace", "Diamonds"),
    
        ("2", "Clubs"), ("3", "Clubs"), ("4", "Clubs"), ("5", "Clubs"), ("6", "Clubs"),
        ("7", "Clubs"), ("8", "Clubs"), ("9", "Clubs"), ("10", "Clubs"),
        ("Jack", "Clubs"), ("Queen", "Clubs"), ("King", "Clubs"), ("Ace", "Clubs"),
    
        ("2", "Spades"), ("3", "Spades"), ("4", "Spades"), ("5", "Spades"), ("6", "Spades"),
        ("7", "Spades"), ("8", "Spades"), ("9", "Spades"), ("10", "Spades"),
        ("Jack", "Spades"), ("Queen", "Spades"), ("King", "Spades"), ("Ace", "Spades")
    ]
    return deck
```

```python
def allPairs(cards):
    for hand in cards:
        if (hand[0][0] != hand[1][0]):
            return False

    return True
```

```python
numTrials = 10000000
hit = 0
results = []

for _ in range(numTrials):
    deck = genDeck()
    random.shuffle(deck)
    cards = []
    for _ in range(2):
        cards.append((deck.pop(0), deck.pop(0)))

    if allPairs(cards):
        hit += 1
        results.append("Hit")
    else:
        results.append("Other")

odds = str(hit/numTrials*100)
print(odds +"%")
```

```python
ax = seaborn.countplot(x = results)

for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:,}',
                (p.get_x() + p.get_width() / 2., height),
                ha='center', va='bottom', fontsize=10, color='black')
ax.text(0.95, 0.95, odds + "%",
        horizontalalignment='right',
        verticalalignment='top',
        transform=ax.transAxes,
        fontsize=12,
        color="red")
ax
```

The empirical probability of drawing a pair 2 times in a row is 0.354% and the theoretical probability is 0.36%
