# Given that the first card dealt in a game of Texas Holdem is a heart, what is the probability that it is an ace? Assuming there are 4 players, the dealing is round robin, and you receive first.

This study will be a prospective, observational study. The sample space I am observing is all of the heart cards in a standard deck so 13. One trial of the simulation is one dealing rotation. I am measuring instances of an ace of heart being dealt, under the circumstances that the card drawn is a heart.

```python
import random, seaborn
```

```python
deck = [
        ("2", "Hearts"), ("3", "Hearts"), ("4", "Hearts"), ("5", "Hearts"), ("6", "Hearts"),
        ("7", "Hearts"), ("8", "Hearts"), ("9", "Hearts"), ("10", "Hearts"),
        ("Jack", "Hearts"), ("Queen", "Hearts"), ("King", "Hearts"), ("Ace", "Hearts")]
```

```python
match = 0
numRuns = 10000000
results = []
for i in range(numRuns):
    if (random.choice(deck) == ("Ace", "Hearts")):
        match += 1
        results.append("Ace of Hearts")
    else:
        results.append("Other Stuff")

odds = str(match/numRuns*100)
odds + "%"
```

```python
100/13
```

```python
ax = seaborn.countplot(x = results)

for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:,}',
                (p.get_x() + p.get_width() / 2., height),
                ha='center', va='bottom', fontsize=10, color='black')
ax.text(0.95, 0.95, odds + "%",
        horizontalalignment='right',
        verticalalignment='top',
        transform=ax.transAxes,
        fontsize=12,
        color="red")
ax
```

The empirical probability of the first card dealt in a game of Texas Holdem is an ace of hearts is 7.689% based off 10 million trials. The theorhetical probability is 1/13 or 7.692%. These two results are very close
