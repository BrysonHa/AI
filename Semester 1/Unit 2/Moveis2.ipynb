# What is the probability that a random movie has a popularity score over 2.0?

Assuming a mean of 5.0 a standard deviation of 2.0 the probability is 93.319%

```python
import requests
import pandas
import random
import time
import seaborn

API_KEY = ""

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "accept": "application/json"
}
```

```python
hits = 0
popScores = []
for i in range(1000):
    print(f"Step: {i}", end="\r", flush=True)
    while True:
        movie_id = random.randint(2, 1250664)
        endpoint = f'https://api.themoviedb.org/3/movie/{movie_id}'
        response = requests.get(endpoint, headers=headers)
        if response.status_code == 200:
            break
    if response.json()["popularity"] > 2.0:
        hits += 1
    popScores.append(response.json()["popularity"])

p = hits/1000
seaborn.histplot(popScores)
```

The empirical estimation of the probability that a random movie has a popularity score over 2.0 is 15.7% while the theorhetical estimation is 93.319%. This large difference is probabaly due to all of the random movies on TMDB with a zero popularity score. Based on the graph it is safe to say this distribution is not normal.
