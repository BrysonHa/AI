# A player has a 20% chance of winning a hand in blackjack. If they play 5 hands, what’s the probability of winning the hand exactly 2 times?

The assumptions I am making is that there is a 20% chance of the player wining a hand when the player does random moves. This is based on experimentation.
The random variable I am measuring is the player winning the hand.
One trial of the simulation is the player playing 5 hands.
A success outcome is the player winning the hand exactly 2/5 times, failure is any other case.

```python
import gymnasium as gym
import matplotlib.pyplot as plt
import ale_py
```

```python
def runGame(i,p):
    print(f"Step, p: {i, p}", end="\r")
    hits = 0

    # Create environment with no rendering
    env = gym.make('Blackjack-v1', render_mode=None)
    observation, info = env.reset()
    run = 0
    total_reward = 0
    for step in range(10):
        # Take random action
        action = env.action_space.sample()
        observation, reward, terminated, truncated, info = env.step(action)
        total_reward += reward

        if reward > 0:
            if run <= 4:
                hits += 1
            run += 1

        # Stop if the game ends
        if terminated or truncated:
            observation, info = env.reset() 
            total_reward = 0

    env.close()

    if hits == 2:
        return (1, hits/5)
    return (0, hits/5)
```

```python
trials = 100000
p = 0
hits = 0
for i in range(trials):
    result, game_hits = runGame(i, p)
    p += result
    hits += game_hits

(p/trials*100, hits/trials)
```

The simulation produced a mean of about 2 wins per five games, consistent with the expected value of one win of 20%. However, the empirical probability of winning exactly two hands, 30.8%, which is higher than the theoretical binomial prediction of 20.48%, suggesting some deviation from ideal random conditions or sampling variation in the simulation.
